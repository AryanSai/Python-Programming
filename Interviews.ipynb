## Decorators

```python
def smart_div(func):
    def wrapper(a, b):
        if a < b:
            a, b = b, a
        return func(a, b)
    return wrapper

@smart_div #div = smart_div(a,b)
def div(a, b):
    print(a / b)

div(2, 4) 
```

## Iterator

```python
l = [1,2,3]

it = iter(l)

print(next(it))
print(next(it))
print(next(it))
```

```python
class EvenNumber:
    def __iter__(self):
        self.n = 0
        return self

    def __next__(self):
        self.n += 2
        return self.n

even = EvenNumber()
itr = iter(even)
print(next(itr))
print(next(itr)) 
```

## Generator

```python
def numbers(n):
    for i in range(n):
        yield i+1

for i in numbers(10):
    print(i)
```

```python
def read_file(file):
    with open(file) as f:
        for line in f:
            yield line

itr = read_file('myfile.txt')

print(next(itr))
print(next(itr))
print(next(itr))
print(next(itr))
```
